## Generate an `X-API-Key` value
It should be copied both to a corresponding environmental variable and climate telemetry sender automation

```kotlin
import java.security.SecureRandom

val chars = "ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789.-_~"
val random = SecureRandom()

val apiKey = (1..16)
    .map { chars[random.nextInt(chars.length)] }
    .joinToString("")

println(apiKey)
// Example output: k8.N2pv9M4qL-b7R
```
